```python
import pandas as pd
import numpy as np
```

```python
data={
    "Student":["A","B","C","D","E","F","G","H","I","J"],
    "Marks":[50,55,60,62,65,68,70,72,75,150]
    }
```

```python
df=pd.DataFrame(data)
print("Original dataset")
print(df)
```

```python
Q1=df["Marks"].quantile(0.25)
print(Q1)
```

```python
Q3=df["Marks"].quantile(0.75)
print(Q3)
```

```python
IQR=Q3-Q1
print(IQR)
```

```python
lower_limit=Q1-1.5*IQR
upper_limit=Q3+1.5*IQR
print(lower_limit)
print(upper_limit)
```

```python
outliers=df[(df["Marks"]<lower_limit)|(df["Marks"]>upper_limit)]
print(outliers)
```

```python
df["Marks_Treated"]=np.where(df["Marks"]>upper_limit,upper_limit,
                             np.where(df["Marks"]<lower_limit,lower_limit,
                                      df["Marks"]
                                     )
                            )
print(df)
```
