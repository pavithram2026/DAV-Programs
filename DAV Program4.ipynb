```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Student" :["A" ,"B", "C","D" ,"E"],
    "Gender" : ["Male","Female","Male","Female","Male"],
    "StudyHours":[5,6,np.nan,4,7],
    "Attendance":[80,85,75,np.nan,90],
    "Marks":[75,82,68,55,np.nan],
    "Result":["Pass","Pass","Pass","Fail","Pass"]
}
```

```python
df=pd.DataFrame(data)
```

```python
print(df)
```

```python
print(df.isnull().sum())
```

```python
df["StudyHours"] =df["StudyHours"].fillna(
    df["StudyHours"].mean()
)
```

```python
print(df)
```

```python
df["Marks"] =df["Marks"].fillna(
    df["StudyHours"].median()
)
```

```python
print(df)
```

```python
df["Attendance"]=df["Attendance"].ffill()
print(df)
```
